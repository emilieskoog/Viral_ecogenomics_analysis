```python
# RPKM weights, one CoverM TSV per environment
import os
import pandas as pd
import numpy as np

BASE_DIR = "/Users/skoog/Desktop/Revisions"

# drop enveloped lineages
DROP_ORDERS = set()        # family only
DROP_FAMILIES = {
    "Pleolipoviridae", "Poxviridae", "Lipothrixviridae", "Globuloviridae", "Bicaudaviridae", "Fuselloviridae",
    "Ascoviridae", "Kitaviridae", "Unclassified Pokkesviricetes", "Asfarviridae", "Unclassified Herpesvirales",
    "Herpesviridae", "Unclassified Lefavirales", "Unclassified Bunyavirales", "Tristromaviridae", "Baculoviridae",
    "Rhabdoviridae", "Unclassified Mononegavirales", "Unclassified Nidovirales", "Alloherpesviridae", "Mymonaviridae",
    "Cystoviridae", "Phenuiviridae", "Nudiviridae", "Unclassified Haloruvirales",
}

RPKM_FILES = {
    # env_key                       (Revisions subfolder,                              CoverM TSV filename)
    "Picturesque_lake":            ("Picturesque_lake_split_by_contig_faa_updated_revisions_exact_besthit",            "Picturesque_viruses_rpkm_mcf0.75.tsv"),
    "Lake_Bitter_1":               ("Lake_Bitter_1_split_by_contig_faa_updated_revisions_exact_besthit",               "Lake_bitter_B1_viruses_rpkm_mcf0.75.tsv"),
    "Cock_soda_lake":              ("Cock_soda_lake_split_by_contig_faa_updated_revisions_exact_besthit",              "Cock_soda_lake_viruses_rpkm_mcf0.75.tsv"),
    "Lake_tanatar_TC":             ("Lake_tanatar_TC_split_by_contig_faa_updated_revisions_exact_besthit",             "Lake_tanatar_TC_viruses_rpkm_mcf0.75.tsv"),
    "Tanatar_5":                   ("Tanatar_5_split_by_contig_faa_updated_revisions_exact_besthit",                   "Tanatar_5_viruses_rpkm_mcf0.75.tsv"),
    "Cueva_de_la_mora_lake_11m":   ("Cueva_de_la_mora_lake_11m_split_by_contig_faa_updated_revisions_exact_besthit",   "Cueva_de_la_mora_lake_11m_viruses_rpkm_mcf0.75.tsv"),
    "Cueva_de_la_mora_lake_35m":   ("Cueva_de_la_mora_lake_35m_split_by_contig_faa_updated_revisions_exact_besthit",   "Cueva_de_la_mora_lake_35m_viruses_rpkm_mcf0.75.tsv"),
    "Cueva_de_la_mora_lake_3m":    ("Cueva_de_la_mora_lake_3m_split_by_contig_faa_updated_revisions_exact_besthit",    "Cueva_de_la_mora_lake_3m_viruses_rpkm_mcf0.75.tsv"),
    "Lost_city":                   ("Lost_city_split_by_contig_faa_updated_revisions_exact_besthit",                   "Lost_city_viruses_rpkm_mcf0.75.tsv"),
    # freshwater lakes
    "Malaren_B":                   ("Malaren_B_split_by_contig_faa_updated_revisions_exact_besthit",                   "Malaren_B_viruses_rpkm_mcf0.75.tsv"),
    "Lake_Victoria":               ("Lake_Victoria_split_by_contig_faa_updated_revisions_exact_besthit",               "Lake_Victoria_lake_viruses_rpkm_mcf0.75.tsv"),
    "Lake_Soyang":                 ("Lake_Soyang_viruses_split_by_contig_faa_besthit",                                 "Lake_Soyang_viruses_rpkm_mcf0.75.tsv"),
}

# taxonomy summary filename differs (virsorter2)
GENOMAD_VS2_SUMMARY = "all_checkv_quality_filtered_typical_atypical_dereplicated_viruses_from_genomad_virsorter2_virus_summary.tsv"

# freshwater metadata
FRESHWATER_META = {
    # measured values
    "Malaren B":     {"Environment_Type": "Freshwater", "pH": 7.75, "Salinity": 0.10},
    "Lake Victoria": {"Environment_Type": "Freshwater", "pH": 7.40, "Salinity": 0.05},
    "Lake Soyang":   {"Environment_Type": "Freshwater", "pH": 6.90, "Salinity": 0.04},
}

def load_rpkm(env_key):
    """Read a CoverM RPKM TSV, return [contig_id, rpkm]."""
    folder, fname = RPKM_FILES[env_key]
    path = os.path.join(BASE_DIR, folder, fname)
    try:
        t = pd.read_csv(path, sep="\t")
    except FileNotFoundError:
        print(f"[RPKM] WARNING: not found for {env_key}: {path}  -> weights default to 1.0")
        return pd.DataFrame({"contig_id": [], "rpkm": []})
    id_col = t.columns[0]
    rpkm_cols = [c for c in t.columns if "rpkm" in c.lower()]
    val_col = rpkm_cols[0] if rpkm_cols else t.select_dtypes("number").columns[0]
    out = t[[id_col, val_col]].copy()
    out.columns = ["contig_id", "rpkm"]
    out["contig_id"] = out["contig_id"].astype(str)
    # strip .faa to match pI side
    out["contig_id"] = out["contig_id"].str.replace(".faa", "", regex=False)
    n = len(out)
    print(f"[RPKM] {env_key}: loaded {n} contig weights from {fname}" if n else f"[RPKM] {env_key}: empty")
    return out

def wquantile(values, weights, q):
    """Weighted quantile; q=0.5 -> weighted median. Equal weights -> ordinary quantile."""
    import numpy as np
    v = np.asarray(values, float); w = np.asarray(weights, float)
    m = np.isfinite(v) & np.isfinite(w) & (w > 0); v, w = v[m], w[m]
    if v.size == 0: return float("nan")
    o = np.argsort(v); v, w = v[o], w[o]
    cw = np.cumsum(w) - 0.5 * w; cw /= w.sum()
    return float(np.interp(q, cw, v))
```

```python
# WATL median pI from split_by_contig_faa
# canon(): strip .faa and _provirus_<start>_<end>
import os, glob, re, statistics
import pandas as pd
from Bio import SeqIO
from Bio.SeqUtils.ProtParam import ProteinAnalysis

def canon(x):
    x = str(x)
    if x.endswith(".faa"):
        x = x[:-4]
    return re.sub(r"_provirus_\d+_\d+$", "", x)

WATL_FAA_DIR = "/Users/skoog/Desktop/Revisions/WATL_wet_dry_viruses_split_by_contig_faa_updated_revisions_besthit"
# WATL catalog taxonomy
WATL_SUMMARY = "/Users/skoog/Desktop/Revisions/WATL_wet_dry_viruses_split_by_contig_faa_updated_revisions_besthit/WATL_wet_dry_viruses_uniqueIDs_virus_summary.tsv"

# median pI per contig
rows = []
for fp in glob.glob(os.path.join(WATL_FAA_DIR, "*.faa")):
    pis = []
    for rec in SeqIO.parse(fp, "fasta"):
        seq = str(rec.seq)
        if seq:
            try:
                pis.append(ProteinAnalysis(seq).isoelectric_point())
            except Exception:
                pass
    rows.append([os.path.basename(fp), statistics.median(pis) if pis else None, len(pis)])

WATL_pI_df = pd.DataFrame(rows, columns=["File", "Median_pI", "n_structural"])
WATL_pI_df["contig_id"] = WATL_pI_df["File"].map(canon)
WATL_pI_df = WATL_pI_df.drop(columns="File")
WATL_pI_df["name_full"] = "WATL brine"
print(f"WATL pI: {WATL_pI_df['contig_id'].nunique()} contigs, "
      f"{int(WATL_pI_df['Median_pI'].notna().sum())} with a median pI")

# taxonomy
try:
    tax = pd.read_csv(WATL_SUMMARY, sep="\t")
    tax.rename(columns={tax.columns[0]: "contig_id"}, inplace=True)
    tax["contig_id"] = tax["contig_id"].map(canon)
    if "taxonomy" in tax.columns:
        parts = tax["taxonomy"].str.split(";", expand=True)
        for i, lvl in enumerate(["virus", "realm", "kingdom", "phylum", "class", "order", "family"]):
            tax[lvl] = parts[i] if i in parts.columns else None
        def _fill(row, primary, fallbacks):
            v = row.get(primary)
            if pd.notna(v) and str(v).strip():
                return v
            for lv in fallbacks:
                u = row.get(lv)
                if pd.notna(u) and str(u).strip():
                    return f"Unclassified {u}"
            return "Unclassified"
        tax["family"] = tax.apply(lambda r: _fill(r, "family", ["order", "class", "phylum", "kingdom", "realm"]), axis=1)
        tax["order"]  = tax.apply(lambda r: _fill(r, "order",  ["class", "phylum", "kingdom", "realm"]), axis=1)
    WATL_pI_tax_df = WATL_pI_df.merge(tax, on="contig_id", how="left")
    nfam = int(WATL_pI_tax_df["family"].notna().sum()) if "family" in WATL_pI_tax_df else 0
    print(f"WATL pI+tax: {len(WATL_pI_tax_df)} rows; {nfam} with a family assignment")
    if nfam == 0:
        print("  !! 0 families matched -> WATL_SUMMARY is keyed on the OLD ids, not WATLvirus_*. "
              "Point it at a WATLvirus-keyed taxonomy source.")
except FileNotFoundError:
    print(f"  WATL_SUMMARY not found ({WATL_SUMMARY}) -> set the path; pI is still computed above.")
    WATL_pI_tax_df = WATL_pI_df.copy()

```

```python
# environment-level pI stats
import numpy as np, pandas as pd
from scipy.stats import kruskal

N_BOOT      = 1000
MIN_GROUP_N = 30        # min group size for the omnibus
SEED        = 0
OUT_CSV     = "/Users/skoog/Desktop/Revisions/env_pI_descriptive_and_omnibus.csv"

try:
    wquantile
except NameError:
    def wquantile(v, w, q):
        v = np.asarray(v, float); w = np.asarray(w, float)
        m = np.isfinite(v) & np.isfinite(w) & (w > 0)
        v, w = v[m], w[m]
        if v.size == 0: return np.nan
        o = np.argsort(v); v, w = v[o], w[o]
        cw = np.cumsum(w) - 0.5 * w
        cw /= w.sum()
        return float(np.interp(q, cw, v))

# one pI per genome per environment
d = filtered_All_envs_df_with_metadata.copy()
d["Median_pI"] = pd.to_numeric(d["Median_pI"], errors="coerce")
d["Environment_Bin"] = d["Environment_Bin"].astype(str).str.strip()
if "rpkm" not in d.columns:
    d["rpkm"] = 1.0
d["rpkm"] = pd.to_numeric(d["rpkm"], errors="coerce")
d.loc[d["rpkm"] <= 0, "rpkm"] = np.nan
d = d[(d["Environment_Bin"] != "Unclassified") & d["Median_pI"].notna()]

# genome RPKM summed over samples
key = ["contig_id", "Environment_Bin"]
gsum = (d.groupby(key, as_index=False)
          .agg(Median_pI=("Median_pI", "first"),
               rpkm=("rpkm", "sum")))
gsum["rpkm"] = gsum["rpkm"].fillna(0.0)
gsum.loc[gsum["rpkm"] <= 0, "rpkm"] = np.nan

groups   = {e: g["Median_pI"].to_numpy() for e, g in gsum.groupby("Environment_Bin")}
weights  = {e: g["rpkm"].to_numpy()      for e, g in gsum.groupby("Environment_Bin")}
sizes    = {e: len(v) for e, v in groups.items()}

# descriptive table
rows = []
for e in sorted(sizes, key=lambda x: -sizes[x]):
    v = groups[e]; w = weights[e]
    w_ok = np.isfinite(w) & (w > 0)
    ww = w if w_ok.any() else np.ones_like(v)
    rows.append(dict(
        environment=e, n_genomes=sizes[e],
        median_pI_weighted=round(wquantile(v, ww, 0.50), 3),
        q1_weighted=round(wquantile(v, ww, 0.25), 3),
        q3_weighted=round(wquantile(v, ww, 0.75), 3),
        median_pI_unweighted=round(float(np.median(v)), 3),
        pct_rpkm_weighted=round(100 * w_ok.mean(), 1),
    ))
desc = pd.DataFrame(rows)
print("PER-ENVIRONMENT DESCRIPTIVE pI  (copy medians into the text):")
print(desc.to_string(index=False))

# raw KW
allv = list(groups.values())
H0, p0 = kruskal(*allv)
n_all, k_all = sum(sizes.values()), len(groups)
#eps0 = (H0 - k_all + 1) / (n_all - k_all)
#es[b] = (H - k + 1) / (target_n * k - k)      # →  es[b] = H / (target_n * k - 1)
eps0 = H0 / (n_all - 1)
print(f"\nRAW KW (unbalanced, reported only as the inflated reference): "
      f"H={H0:.1f}, p={p0:.1e}, eps^2={eps0:.4f}, k={k_all}")

# balanced omnibus
retained = [e for e in groups if sizes[e] >= MIN_GROUP_N]
excluded = [e for e in groups if sizes[e] <  MIN_GROUP_N]
k = len(retained)
n_min = min(sizes[e] for e in retained)
n_med = int(round(np.median([sizes[e] for e in retained])))
print(f"\nomnibus groups (n>={MIN_GROUP_N}): {retained}")
if excluded:
    print(f"excluded from omnibus (too small, kept descriptively): "
          f"{[(e, sizes[e]) for e in excluded]}")

def balanced_kw(target_n, seed=SEED):
    rng = np.random.default_rng(seed)
    Hs = np.empty(N_BOOT); ps = np.empty(N_BOOT); es = np.empty(N_BOOT)
    for b in range(N_BOOT):
        samp = []
        for e in retained:
            v = groups[e]
            replace = target_n > len(v)
            samp.append(rng.choice(v, size=target_n, replace=replace))
        H, p = kruskal(*samp)
        Hs[b], ps[b] = H, p
        es[b] = H / (target_n * k - 1)
        #es[b] = (H - k + 1) / (target_n * k - k)
    ci = lambda a: (np.percentile(a, 2.5), np.median(a), np.percentile(a, 97.5))
    return Hs, ps, es, ci

for label, tn in [("min-of-retained (headline, no up-sampling)", n_min),
                  ("median-of-retained (sensitivity)",           n_med)]:
    Hs, ps, es, ci = balanced_kw(tn)
    hl, hm, hh = ci(Hs); el, em, eh = ci(es)
    print(f"\nBALANCED KW  [{label}]  n={tn}/group x {N_BOOT} boot, {k} groups:")
    print(f"  H        median={hm:7.1f}   95% CI [{hl:.1f}, {hh:.1f}]")
    print(f"  p        median={np.median(ps):.2e}   frac p<0.05={np.mean(ps<0.05):.3f}")
    print(f"  eps^2    median={em:.4f}   95% CI [{el:.4f}, {eh:.4f}]"
          f"   (rank-based: ~0.01 small / 0.08 medium / 0.26 large)")

desc.to_csv(OUT_CSV, index=False)
print(f"\nwrote {OUT_CSV}")
```

```python
# pairwise comparisons (panel D)- OLD ANALYSIS
import numpy as np, pandas as pd
import matplotlib.pyplot as plt, seaborn as sns
from scipy.stats import mannwhitneyu
from itertools import combinations

N_BOOT      = 1000
MIN_GROUP_N = 30
ORDER_BY    = "median_pI"   # "median_pI" | "salinity"
SEED        = 0
OUT_PREFIX  = "/Users/skoog/Desktop/panelD_pairwise_pI"

# data
d = filtered_All_envs_df_with_metadata.copy()
d["Median_pI"] = pd.to_numeric(d["Median_pI"], errors="coerce")
d["Environment_Bin"] = d["Environment_Bin"].astype(str).str.strip()
d = d[(d["Environment_Bin"] != "Unclassified") & d["Median_pI"].notna()]
g1 = d.drop_duplicates(["contig_id", "Environment_Bin"])
groups = {e: sub["Median_pI"].to_numpy() for e, sub in g1.groupby("Environment_Bin")}
sizes  = {e: len(v) for e, v in groups.items()}
small  = {e for e in sizes if sizes[e] < MIN_GROUP_N}

if ORDER_BY == "salinity" and "Salinity" in g1.columns and "name_full" in g1.columns:
    sal = (g1.groupby("Environment_Bin")
             .apply(lambda x: x.groupby("name_full")["Salinity"].first().mean()))
    envs = sal.sort_values().index.tolist()
else:
    envs = (g1.groupby("Environment_Bin")["Median_pI"].median()
              .sort_values().index.tolist())

def cliffs(a, b):
    U, _ = mannwhitneyu(a, b, alternative="two-sided")
    return 2.0 * U / (len(a) * len(b)) - 1.0

def bh(p):
    p = np.asarray(p, float); n = len(p); o = np.argsort(p)
    adj = np.minimum.accumulate((p[o] * n / np.arange(1, n + 1))[::-1])[::-1]
    out = np.empty(n); out[o] = np.clip(adj, 0, 1); return out

rng = np.random.default_rng(SEED)
delta = pd.DataFrame(np.nan, index=envs, columns=envs)
diff  = pd.DataFrame(np.nan, index=envs, columns=envs)
pairs, raw_p, flags = list(combinations(envs, 2)), [], []
for a, b in pairs:
    A, B = groups[a], groups[b]
    tn = min(len(A), len(B))
    ps = np.empty(N_BOOT)
    for i in range(N_BOOT):
        sa = rng.choice(A, tn, replace=(tn > len(A)))
        sb = rng.choice(B, tn, replace=(tn > len(B)))
        ps[i] = mannwhitneyu(sa, sb, alternative="two-sided")[1]
    dd, md = cliffs(A, B), float(np.median(A) - np.median(B))
    delta.loc[a, b], delta.loc[b, a] = dd, -dd
    diff.loc[a, b],  diff.loc[b, a]  = md, -md
    raw_p.append(float(np.median(ps)))
    flags.append((a in small) or (b in small))

padj = dict(zip(pairs, bh(raw_p)))
def stars(p): return "***" if p < .001 else "**" if p < .01 else "*" if p < .05 else "ns"

# table
def _mag(x):
    a = abs(x)
    return "negligible" if a < .147 else "small" if a < .33 else "medium" if a < .474 else "large"
tab = pd.DataFrame({
    "group_1": [a for a, b in pairs], "group_2": [b for a, b in pairs],
    "n1": [sizes[a] for a, b in pairs], "n2": [sizes[b] for a, b in pairs],
    "median_pI_diff": [round(diff.loc[a, b], 2) for a, b in pairs],
    "cliffs_delta":   [round(delta.loc[a, b], 3) for a, b in pairs],
    "magnitude":      [_mag(delta.loc[a, b]) for a, b in pairs],
    "q_fdr":          [padj[(a, b)] for a, b in pairs],
    "low_confidence": flags,
}).reindex(pd.Series([abs(delta.loc[a, b]) for a, b in pairs]).sort_values(ascending=False).index)
print("PAIRWISE pI (sorted by |Cliff's delta|; low_confidence = involves a group n<%d):" % MIN_GROUP_N)
print(tab.to_string(index=False))
tab.to_csv(OUT_PREFIX + ".csv", index=False)

# Cliff's delta heatmap
annot = pd.DataFrame("", index=envs, columns=envs)
for a, b in pairs:
    s = stars(padj[(a, b)])
    annot.loc[a, b] = f"{diff.loc[a,b]:+.2f}\n{s}"
    annot.loc[b, a] = f"{diff.loc[b,a]:+.2f}\n{s}"
plt.figure(figsize=(1.4 * len(envs) + 3, 1.1 * len(envs) + 2))
ax = sns.heatmap(delta.astype(float), cmap="vlag", center=0, vmin=-1, vmax=1,
                 linewidths=.5, linecolor="white", mask=delta.isna(),
                 annot=annot.values, fmt="", annot_kws={"size": 7},
                 cbar_kws={"label": "Cliff's delta (row vs col)"})
ax.set_title("Pairwise structural-pI difference across environments\n"
             f"per-pair balanced bootstrap, FDR-corrected  (* .05  ** .01  *** .001)",
             fontsize=11, pad=10)
ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha="right")
plt.tight_layout()
plt.savefig(OUT_PREFIX + "_cliffs.svg", bbox_inches="tight", transparent=True)
plt.savefig(OUT_PREFIX + "_cliffs.png", dpi=300, bbox_inches="tight")
plt.show()

# -log10(q) heatmap
qmat = pd.DataFrame(1.0, index=envs, columns=envs)
for (a, b), q in padj.items():
    qmat.loc[a, b] = q; qmat.loc[b, a] = q
neglogq = -np.log10(qmat.clip(lower=1e-300))
mask = np.triu(np.ones(neglogq.shape, dtype=bool), k=1)
sig = pd.DataFrame("", index=envs, columns=envs)
for (a, b), q in padj.items():
    st = "***" if q < .001 else "**" if q < .01 else "*" if q < .05 else ""
    sig.loc[a, b] = st; sig.loc[b, a] = st
plt.figure(figsize=(1.4 * len(envs) + 3, 1.1 * len(envs) + 2))
ax = sns.heatmap(neglogq.astype(float), cmap="mako", mask=mask | neglogq.isna(),
                 linewidths=.5, linecolor="white", annot=sig.values, fmt="",
                 annot_kws={"size": 9, "color": "white"},
                 cbar_kws={"label": "Significance  [-log10(q)]"})
ax.set_title("Pairwise structural-pI significance across environments\n"
             "balanced bootstrap, FDR-corrected  (* q<.05  ** q<.01  *** q<.001)",
             fontsize=11, pad=10)
ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha="right")
plt.tight_layout()
plt.savefig(OUT_PREFIX + "_neglogq.svg", bbox_inches="tight", transparent=True)
plt.savefig(OUT_PREFIX + "_neglogq.png", dpi=300, bbox_inches="tight")
plt.show()

n_sig = sum(padj[(a, b)] < 0.05 for a, b in pairs)
print(f"\npanel D: {len(envs)} environments, {n_sig}/{len(pairs)} pairs FDR<0.05; "
      f"max -log10(q)={np.nanmax(neglogq.values):.0f}; "
      f"{sum(flags)} pair(s) flagged low-confidence (small group).")
```

```python
# structural ORF fraction per genome
import os, glob, re
import pandas as pd

def canon(x):
    x = str(x)
    if x.endswith(".faa"):
        x = x[:-4]
    x = x.replace("|", "_")
    return re.sub(r"_provirus_\d+_\d+$", "", x)      # strip provirus suffix

BASE = "/Users/skoog/Desktop/Revisions"
SUMMARY = "all_checkv_quality_filtered_typical_atypical_dereplicated_viruses_from_genomad_virus_summary.tsv"
SUMMARY_VS2 = "all_checkv_quality_filtered_typical_atypical_dereplicated_viruses_from_genomad_virsorter2_virus_summary.tsv"

# WATL
WATL_FAA_DIR         = f"{BASE}/WATL_wet_dry_viruses_split_by_contig_faa_updated_revisions_besthit"
WATL_QUALITY_SUMMARY = f"{BASE}/WATL_wet_dry_viruses_split_by_contig_faa_updated_revisions_besthit/WATL_wet_dry_viruses_uniqueIDs_virus_summary.tsv"

ENV_STRUCT = {
    "WATL brine":                                      (WATL_FAA_DIR, WATL_QUALITY_SUMMARY),
    "Picturesque Lake":                                (f"{BASE}/Picturesque_lake_split_by_contig_faa_updated_revisions_exact_besthit", SUMMARY),
    "Lake Tanatar trona crystallizer":                 (f"{BASE}/Lake_tanatar_TC_split_by_contig_faa_updated_revisions_exact_besthit",  SUMMARY),
    "Lake Tanatar-5":                                  (f"{BASE}/Tanatar_5_split_by_contig_faa_updated_revisions_exact_besthit",        SUMMARY),
    "Lake Bitter-1":                                   (f"{BASE}/Lake_Bitter_1_split_by_contig_faa_updated_revisions_exact_besthit",    SUMMARY),
    "Cueva de la Mora lake - Chemocline (11m)":        (f"{BASE}/Cueva_de_la_mora_lake_11m_split_by_contig_faa_updated_revisions_exact_besthit", SUMMARY),
    "Cueva de la Mora lake - Deep Anoxic Layer (35m)": (f"{BASE}/Cueva_de_la_mora_lake_35m_split_by_contig_faa_updated_revisions_exact_besthit", SUMMARY),
    "Cueva de la Mora lake - Upper Oxic Layer (3m)":   (f"{BASE}/Cueva_de_la_mora_lake_3m_split_by_contig_faa_updated_revisions_exact_besthit",  SUMMARY),
    "Cock Soda Lake":                                  (f"{BASE}/Cock_soda_lake_split_by_contig_faa_updated_revisions_exact_besthit",   SUMMARY),
    "Lost City":                                       (f"{BASE}/Lost_city_split_by_contig_faa_updated_revisions_exact_besthit",        SUMMARY),
    "Malaren B":                                       (f"{BASE}/Malaren_B_split_by_contig_faa_updated_revisions_exact_besthit",       SUMMARY_VS2),
    "Lake Victoria":                                   (f"{BASE}/Lake_Victoria_split_by_contig_faa_updated_revisions_exact_besthit",   SUMMARY_VS2),
    "Lake Soyang":                                     (f"{BASE}/Lake_Soyang_viruses_split_by_contig_faa_besthit",       SUMMARY_VS2),
}

IMGVR_COMBINED_FAA = f"{BASE}/all_checkv_quality_filtered_typical_atypical_dereplicated_viruses_from_genomad_blast_vog_db_3_viral_structural_genes1_contigIDs_merged_viral_structural_genes_dir_updated_revisions_exact_besthit/viral_structural_genes_IMGVR_all_proteins_Freshwater_len3500_proteins.clean.faa"
IMGVR_META         = f"{BASE}/all_checkv_quality_filtered_typical_atypical_dereplicated_viruses_from_genomad_blast_vog_db_3_viral_structural_genes1_contigIDs_merged_viral_structural_genes_dir_updated_revisions_exact_besthit/IMGVR_all_Sequence_information_Freshwater_len3500.tsv"
_UVIG_RE = re.compile(r"(IMGVR_UViG_\d+_\d+)")
def _parse_uvig(h):
    m = _UVIG_RE.search(str(h))
    if m: return m.group(1)
    p = str(h).split("_"); return "_".join(p[:4]) if len(p) > 4 else str(h)
def _n_structural_per_uvig_combined(faa):
    from collections import Counter
    c = Counter()
    for line in open(faa):
        if line.startswith(">"):
            c[_parse_uvig(line[1:].split()[0])] += 1
    return dict(c)

def _n_structural_per_contig(faa_dir):
    out = {}
    for fp in glob.glob(os.path.join(faa_dir, "*.faa")):
        base = os.path.basename(fp)
        if base.endswith(".faa"):
            base = base[:-4]
        full = base.replace("|", "_")
        out[full] = sum(1 for line in open(fp) if line.startswith(">"))
    return out

def _find_ngenes_col(df):
    for c in df.columns:
        if c.lower() in ("n_genes", "ngenes", "gene_count", "num_genes"):
            return c
    return None

def load_imgvr_meta(path):
    m = pd.read_csv(path, sep="\t")
    gene_col = next(c for c in m.columns if "gene content" in c.lower())
    out = pd.DataFrame({"contig_id": m[m.columns[0]].astype(str).map(canon)})
    out["n_genes"] = pd.to_numeric(m[gene_col].astype(str).str.split(";").str[0], errors="coerce")
    return out

def structural_orf_fraction(env_name):
    faa_dir, summ = ENV_STRUCT[env_name]
    summ_path = summ if os.path.isabs(summ) else os.path.join(faa_dir, summ)
    ns = _n_structural_per_contig(faa_dir)
    if not ns:
        print(f"[struct] {env_name}: no .faa in {faa_dir}"); return pd.DataFrame()
    try:
        s = pd.read_csv(summ_path, sep="\t")
    except FileNotFoundError:
        print(f"[struct] {env_name}: summary not found {summ_path}"); return pd.DataFrame()
    ng_col = _find_ngenes_col(s)
    if ng_col is None:
        print(f"[struct] {env_name}: no gene-count column (cols: {list(s.columns)[:8]}) -> skipped")
        return pd.DataFrame()
    s = s[[s.columns[0], ng_col]].copy(); s.columns = ["stripped_id", "n_genes"]
    s["stripped_id"] = s["stripped_id"].astype(str).map(canon)
    frac = pd.DataFrame({"contig_id": list(ns), "n_structural_orfs": list(ns.values())})
    frac["stripped_id"] = frac["contig_id"].map(canon)
    frac = frac.merge(s, on="stripped_id", how="left").drop(columns="stripped_id")
    frac["n_genes"] = pd.to_numeric(frac["n_genes"], errors="coerce")
    frac["struct_orf_fraction"] = frac["n_structural_orfs"] / frac["n_genes"]
    frac["name_full"] = env_name
    n_anom = int((frac["n_structural_orfs"] > frac["n_genes"]).sum())
    msg = f"[struct] {env_name}: {len(frac)} genomes, median fraction = {frac['struct_orf_fraction'].median():.3f}"
    if n_anom: msg += f"  (!! {n_anom} with n_structural>n_genes -> check ID/ORF-caller match)"
    print(msg)
    return frac

def structural_orf_fraction_imgvr():
    try:
        ns = _n_structural_per_uvig_combined(IMGVR_COMBINED_FAA)
    except FileNotFoundError:
        print(f"[struct] Freshwater (IMG): combined faa not found {IMGVR_COMBINED_FAA}"); return pd.DataFrame()
    if not ns:
        print(f"[struct] Freshwater (IMG): no proteins parsed from {IMGVR_COMBINED_FAA}"); return pd.DataFrame()
    try:
        meta = load_imgvr_meta(IMGVR_META)
    except FileNotFoundError:
        print(f"[struct] Freshwater (IMG): metadata not found {IMGVR_META}"); return pd.DataFrame()
    frac = pd.DataFrame({"contig_id": list(ns), "n_structural_orfs": list(ns.values())}).merge(meta, on="contig_id", how="left")
    frac["struct_orf_fraction"] = frac["n_structural_orfs"] / frac["n_genes"]
    frac["name_full"] = "Freshwater (IMG)"
    print(f"[struct] Freshwater (IMG): {len(frac)} genomes, median fraction = {frac['struct_orf_fraction'].median():.3f}")
    return frac

struct_frac_all = pd.concat(
    [structural_orf_fraction(e) for e in ENV_STRUCT],
    ignore_index=True)
struct_frac_summary = (struct_frac_all.groupby("name_full")["struct_orf_fraction"]
                       .agg(["count", "median", "mean", "std"]).reset_index())
display(struct_frac_summary)
struct_frac_all.to_csv(f"{BASE}/structural_orf_fraction_per_genome.csv", index=False)
print("wrote structural_orf_fraction_per_genome.csv")
```

```python
# environment-type comparisons, lake as unit
import warnings
warnings.filterwarnings("ignore", module="statsmodels")
import re
import numpy as np, pandas as pd
import matplotlib.pyplot as plt, seaborn as sns
from scipy.stats import kruskal, mannwhitneyu
import statsmodels.formula.api as smf
from scipy.stats import chi2
from statsmodels.stats.multitest import multipletests
import math
from itertools import combinations

MIN_LAKES  = 3
WEIGHTED   = True     # True = RPKM-weighted
PI         = "pI_weighted" if WEIGHTED else "pI_unweighted"
PANEL_D_TEST = "hboot"   # "hboot" or "mwu_lakes"
N_HBOOT    = 10000
SEED       = 3
OUT_PREFIX = "/Users/skoog/Desktop/Revisions/panelD_LAKE_UNIT"
LAKE_ALIASES = {"Sunrise Hill": "Sunrise Hill Lake"}
# Gounter 3A/3B and Twin Lake West 7B are considered separate lakes
CUEVA_KEY = "Cueva de la Mora"

def physical_lake(name):
    s = re.sub(r"\s+", " ", str(name)).strip()
    if CUEVA_KEY.lower() in s.lower(): return CUEVA_KEY
    return LAKE_ALIASES.get(s, s)

d = filtered_All_envs_df_with_metadata.copy()
d["Median_pI"] = pd.to_numeric(d["Median_pI"], errors="coerce")
d["rpkm"] = pd.to_numeric(d.get("rpkm", np.nan), errors="coerce")
d["Environment_Bin"] = d["Environment_Bin"].astype(str).str.strip()
d = d[(d["Environment_Bin"] != "Unclassified") & d["Median_pI"].notna()].copy()
d["Physical_lake"] = d["name_full"].map(physical_lake)

# lakes split across pH classes count once in each (Sunrise Hill)
switch = d.groupby("Physical_lake")["Environment_Bin"].nunique()
SPLIT_LAKES = list(switch[switch > 1].index)
for lk in SPLIT_LAKES:
    print(f"note: {lk} spans {sorted(d.loc[d.Physical_lake == lk, 'Environment_Bin'].unique())} "
          f"-> one unit in each class")
_split = d["Physical_lake"].isin(SPLIT_LAKES)
d.loc[_split, "Physical_lake"] = (d.loc[_split, "Physical_lake"] + " ["
                                  + d.loc[_split, "Environment_Bin"].str.replace(r"\s*lakes \(WATL\)", "", regex=True) + "]")
d["Env_lake"] = d["Environment_Bin"]

# genome x lake
gk = (d.groupby(["Env_lake", "Physical_lake", "contig_id"], as_index=False)
        .agg(pI=("Median_pI", "first"), rpkm=("rpkm", lambda s: s.sum(min_count=1))))

def _wmed(v, w):
    v = np.asarray(v, float); w = np.asarray(w, float)
    ok = np.isfinite(v) & np.isfinite(w) & (w > 0)
    if not ok.any(): return float(np.median(v))
    v, w = v[ok], w[ok]; o = np.argsort(v); v, w = v[o], w[o]
    return float(np.interp(0.5, (np.cumsum(w) - 0.5 * w) / w.sum(), v))

lakes = (gk.groupby(["Env_lake", "Physical_lake"])
           .apply(lambda x: pd.Series({"pI_unweighted": x["pI"].median(),
                                       "pI_weighted": _wmed(x["pI"], x["rpkm"]),
                                       "n_genomes": x["contig_id"].nunique()}))
           .reset_index())
per_type = (lakes.groupby("Env_lake")
                 .agg(n_lakes=("Physical_lake", "nunique"), n_genomes=("n_genomes", "sum"),
                      median_of_lake_medians=(PI, "median"),
                      min_lake=(PI, "min"), max_lake=(PI, "max"))
                 .reset_index())
print("\nLakes per environment type:")
print(per_type.round(3).to_string(index=False))

multi  = per_type.loc[per_type["n_lakes"] >= MIN_LAKES, "Env_lake"].tolist()
single = per_type.loc[per_type["n_lakes"] <  MIN_LAKES, "Env_lake"].tolist()
print(f"\ntested (>= {MIN_LAKES} lakes): {multi}\ndescriptive only: {single}")

# omnibus
omni = []
for col in (PI, "pI_unweighted" if WEIGHTED else "pI_weighted"):
    grp = [lakes.loc[lakes.Env_lake == e, col].to_numpy() for e in multi]
    H, p = kruskal(*grp); n = sum(len(g) for g in grp)
    omni.append(dict(test=f"Kruskal-Wallis on lake medians ({col})", stat=H, df=len(grp) - 1,
                     p=p, eps2=H / (n - 1), n_lakes=n))
gm = gk[gk["Env_lake"].isin(multi)].copy()
try:
    m1 = smf.mixedlm("pI ~ C(Env_lake)", gm, groups=gm["Physical_lake"]).fit(reml=False)
    m0 = smf.mixedlm("pI ~ 1",            gm, groups=gm["Physical_lake"]).fit(reml=False)
    lr = 2 * (m1.llf - m0.llf); dfm = len(multi) - 1
    omni.append(dict(test="LMM genome pI ~ type + (1|lake), LRT", stat=lr, df=dfm,
                     p=chi2.sf(lr, dfm), eps2=np.nan, n_lakes=gm["Physical_lake"].nunique()))
except Exception as e:
    print("mixed model failed:", e)
omni = pd.DataFrame(omni)
print("\nOMNIBUS (lake as unit):"); print(omni.round(5).to_string(index=False))

# pairwise, hierarchical bootstrap
rng = np.random.default_rng(SEED)
def _w(x):
    return np.nan_to_num(np.asarray(x, float), nan=0.0) if WEIGHTED else np.ones(len(x))
pools = {e: [(g["pI"].to_numpy(float), _w(g["rpkm"])) for _, g in gk[gk.Env_lake == e].groupby("Physical_lake")]
         for e in multi}
def _hboot_median(lake_arrays):
    """Resample lakes, then genomes within lakes; weighted median."""
    pick = rng.integers(0, len(lake_arrays), len(lake_arrays))
    vs, ws = [], []
    for i in pick:
        v, w = lake_arrays[i]; idx = rng.integers(0, len(v), len(v))
        vs.append(v[idx]); ws.append(w[idx])
    return _wmed(np.concatenate(vs), np.concatenate(ws))
def _pooled_median(lake_arrays):
    return _wmed(np.concatenate([v for v, _ in lake_arrays]), np.concatenate([w for _, w in lake_arrays]))
B = {e: np.array([_hboot_median(pools[e]) for _ in range(N_HBOOT)]) for e in multi}

def cliffs(a, b):
    U = mannwhitneyu(a, b, alternative="two-sided").statistic
    return 2.0 * U / (len(a) * len(b)) - 1.0

rows = []
for a, b in combinations(multi, 2):
    diff = B[a] - B[b]
    p_boot = min(1.0, 2 * min((diff <= 0).mean(), (diff >= 0).mean()))
    p_boot = max(p_boot, 1.0 / N_HBOOT)
    la = lakes.loc[lakes.Env_lake == a, PI].to_numpy()
    lb = lakes.loc[lakes.Env_lake == b, PI].to_numpy()
    rows.append(dict(group_1=a, group_2=b, n_lakes_1=len(la), n_lakes_2=len(lb),
                     pooled_median_diff=_pooled_median(pools[a]) - _pooled_median(pools[b]),
                     diff_ci_lo=np.percentile(diff, 2.5), diff_ci_hi=np.percentile(diff, 97.5),
                     p_hboot=p_boot, cliffs_delta_lakes=cliffs(la, lb),
                     p_mwu_lakes=mannwhitneyu(la, lb, alternative="two-sided").pvalue,
                     min_attainable_p_mwu=2.0 / math.comb(len(la) + len(lb), len(la))))
pw = pd.DataFrame(rows)
pw["q_hboot"] = multipletests(pw["p_hboot"], method="fdr_bh")[1]
pw["q_mwu_lakes"] = multipletests(pw["p_mwu_lakes"], method="fdr_bh")[1]
print("\nPAIRWISE (hierarchical bootstrap; lake-level Cliff's delta / exact MWU):")
print("CAUTION: with < ~5 lakes per type the cluster bootstrap understates between-lake variance;"
      "\n         where min_attainable_p_mwu > 0.05 report the pair as underpowered, not significant.")
print(pw.round(4).to_string(index=False))

# split lakes excluded
if SPLIT_LAKES:
    lx = lakes[~lakes["Physical_lake"].str.contains("|".join(map(re.escape, SPLIT_LAKES)))]
    g_ = [lx.loc[lx.Env_lake == e, PI].to_numpy() for e in multi]
    H_, p_ = kruskal(*g_)
    ps_ = [mannwhitneyu(lx.loc[lx.Env_lake == a, PI], lx.loc[lx.Env_lake == b, PI],
                        alternative="two-sided").pvalue for a, b in combinations(multi, 2)]
    sens = pd.DataFrame({"group_1": [a for a, b in combinations(multi, 2)],
                         "group_2": [b for a, b in combinations(multi, 2)],
                         "q_mwu_lakes_without_split": multipletests(ps_, method="fdr_bh")[1]})
    print(f"\nSENSITIVITY without {SPLIT_LAKES}: Kruskal-Wallis H = {H_:.2f}, p = {p_:.4f}")
    print(sens.merge(pw[["group_1", "group_2", "q_mwu_lakes"]]).round(4).to_string(index=False))

# single-site types
desc = []
for e in single:
    ge = gk[gk.Env_lake == e]; v = ge["pI"].to_numpy(float); w = _w(ge["rpkm"])
    def _wq(q):
        ok = w > 0
        if not ok.any(): return float(np.quantile(v, q))
        o = np.argsort(v[ok]); vv, ww = v[ok][o], w[ok][o]
        return float(np.interp(q, (np.cumsum(ww) - 0.5 * ww) / ww.sum(), vv))
    row = dict(environment=e, sites=", ".join(sorted(ge["Physical_lake"].unique())),
               n_genomes=len(v), median_pI=_wq(.5), q1=_wq(.25), q3=_wq(.75))
    for m in multi:
        lm = lakes.loc[lakes.Env_lake == m, PI]
        row[f"percentile_among_{m}_lakes"] = 100 * (lm < row["median_pI"]).mean()
    desc.append(row)
desc = pd.DataFrame(desc)
print("\nSINGLE-SITE TYPES (descriptive):"); print(desc.round(3).to_string(index=False))

with pd.ExcelWriter(OUT_PREFIX + "_tables.xlsx", engine="openpyxl") as xw:
    for nm, t in [("lakes", lakes), ("per_type", per_type), ("omnibus", omni),
                  ("pairwise", pw), ("single_site", desc)]:
        t.to_excel(xw, sheet_name=nm, index=False)
        ws = xw.sheets[nm]; ws.auto_filter.ref = ws.dimensions

# panel D
from matplotlib.colors import to_hex
NT_LABEL = ""
NT_COLOR = "0.88"
VMAX     = None

envs_all = (globals().get("envs") or
            d.drop_duplicates(["contig_id", "Environment_Bin"]).groupby("Environment_Bin")["Median_pI"]
             .median().sort_values().index.tolist())
envs_all = [e for e in envs_all if e in set(per_type["Env_lake"])]

def wcliffs(x, wx, y, wy):
    """Abundance-weighted Cliff's delta: P_w(X > Y) - P_w(X < Y)."""
    x, y = np.asarray(x, float), np.asarray(y, float)
    wx, wy = np.asarray(wx, float), np.asarray(wy, float)
    if wx.sum() <= 0: wx = np.ones_like(x)
    if wy.sum() <= 0: wy = np.ones_like(y)
    o = np.argsort(y); y, wy = y[o], wy[o]; cw = np.concatenate([[0.0], np.cumsum(wy)])
    less = cw[np.searchsorted(y, x, side="left")]
    greater = cw[-1] - cw[np.searchsorted(y, x, side="right")]
    return float(np.sum(wx * (less - greater)) / (wx.sum() * wy.sum()))

if WEIGHTED:
    # genome x environment type
    ge_ = (d.groupby(["Environment_Bin", "contig_id"], as_index=False)
             .agg(pI=("Median_pI", "first"), rpkm=("rpkm", lambda s: s.sum(min_count=1))))
    gr_ = {e: (x["pI"].to_numpy(float), _w(x["rpkm"])) for e, x in ge_.groupby("Environment_Bin")}
    dg = pd.DataFrame(np.nan, index=envs_all, columns=envs_all)
    for a_, b_ in combinations(envs_all, 2):
        v = wcliffs(*gr_[a_], *gr_[b_]); dg.loc[a_, b_], dg.loc[b_, a_] = v, -v
elif "delta" in globals() and set(envs_all) <= set(delta.index):
    dg = delta.loc[envs_all, envs_all].astype(float)
else:
    g1_ = d.drop_duplicates(["contig_id", "Environment_Bin"])
    grp_ = {e: x["Median_pI"].to_numpy() for e, x in g1_.groupby("Environment_Bin")}
    dg = pd.DataFrame(np.nan, index=envs_all, columns=envs_all)
    for a_, b_ in combinations(envs_all, 2):
        v = cliffs(grp_[a_], grp_[b_]); dg.loc[a_, b_], dg.loc[b_, a_] = v, -v

qcol = "q_mwu_lakes" if PANEL_D_TEST == "mwu_lakes" else "q_hboot"
qa = pd.DataFrame(np.nan, index=envs_all, columns=envs_all)
fa = pd.DataFrame(False, index=envs_all, columns=envs_all)
for _, r in pw.iterrows():
    qa.loc[r.group_1, r.group_2] = qa.loc[r.group_2, r.group_1] = r[qcol]
    fa.loc[r.group_1, r.group_2] = fa.loc[r.group_2, r.group_1] = r.min_attainable_p_mwu > 0.05

rows_a, cols_a = envs_all[1:], envs_all[:-1]
L = -np.log10(qa.loc[rows_a, cols_a].astype(float).clip(lower=1e-300))
tri = np.triu(np.ones(L.shape, bool), k=1)
cmap = plt.get_cmap("mako")
vmax = VMAX or max(1.5, float(np.nanmax(L.values)) if np.isfinite(L.values).any() else 1.5)
fig, ax = plt.subplots(figsize=(1.6 * len(cols_a) + 2.5, 0.9 * len(rows_a) + 1.6))
sns.heatmap(L, mask=tri | L.isna().values, cmap="mako", vmin=0, vmax=vmax, annot=False,
            linewidths=.5, linecolor="white", cbar_kws={"label": "Significance [-log10(FDR q)]"}, ax=ax)
for i, a_ in enumerate(rows_a):
    for j, b_ in enumerate(cols_a):
        if tri[i, j]: continue
        dv = dg.loc[a_, b_]; q_ = qa.loc[a_, b_]
        if np.isfinite(q_):
            st = "***" if q_ < .001 else "**" if q_ < .01 else "*" if q_ < .05 else ""
            st += "\u2020" if (fa.loc[a_, b_] and PANEL_D_TEST == "mwu_lakes") else ""
            light = L.loc[a_, b_] > 0.6 * vmax
            txt, col = f"{st}\n{dv:+.2f}", ("0.1" if light else "white")
        else:
            txt, col = (f"{NT_LABEL}\n{dv:+.2f}" if NT_LABEL else f"{dv:+.2f}"), "0.35"
        ax.text(j + 0.5, i + 0.5, txt, ha="center", va="center", fontsize=9, color=col)
# untested cells
from matplotlib.patches import Rectangle
for i, a_ in enumerate(rows_a):
    for j, b_ in enumerate(cols_a):
        if not tri[i, j] and not np.isfinite(qa.loc[a_, b_]):
            ax.add_patch(Rectangle((j, i), 1, 1, facecolor=NT_COLOR, lw=0, zorder=2))
for t in ax.texts: t.set_zorder(3)
for k in range(len(cols_a) + 1): ax.axvline(k, color="white", lw=0.5)
for k in range(len(rows_a) + 1): ax.axhline(k, color="white", lw=0.5)
lab = lambda t: t.replace("WATL", "WAHL")
ax.set_xticks(np.arange(len(cols_a)) + 0.5); ax.set_xticklabels([lab(c) for c in cols_a], rotation=45, ha="right")
ax.set_yticks(np.arange(len(rows_a)) + 0.5); ax.set_yticklabels([lab(r) for r in rows_a], rotation=0)
_how = ("exact Mann-Whitney on lake medians" if PANEL_D_TEST == "mwu_lakes"
        else "hierarchical bootstrap: lakes, then genomes within lakes")
_how = ("RPKM-weighted; " if WEIGHTED else "") + _how
_dag = "; \u2020 too few lakes (3 vs 4) for q < 0.05 to be reachable" if PANEL_D_TEST == "mwu_lakes" else ""
ax.set_title(f"FDR-corrected q ({_how}); number = {'RPKM-weighted ' if WEIGHTED else ''}Cliff's \u03b4 (per genome)\n"
             f"grey: not tested (environment type from a single site){_dag}", fontsize=9)
fig.tight_layout()
fig.savefig(OUT_PREFIX + "_all7.svg", bbox_inches="tight"); fig.savefig(OUT_PREFIX + "_all7.png", dpi=300, bbox_inches="tight")
plt.show()
```
